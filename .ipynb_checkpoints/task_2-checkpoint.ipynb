Задание 2

```python
def date_range(date_start, date_end): #функция, 2 значение на вход
    try: #проверяем формат
        date1 = datetime.strptime(date_start, '%Y-%m-%d')
        date2 = datetime.strptime(date_end, '%Y-%m-%d')
    except ValueError: #в случае неверного формата возвращаем пустой список
        return []
    if date1 > date2: #если дата 1 > дата 2 также возвращаем пустой список
        return []
    else:
        list = [date_start] #генерируем список
        for i in range((date2-date1).days):
            date1 += timedelta(days=1)
            list.append(date1.strftime('%Y-%m-%d'))
        return(list)
```

```python
date_range('2022-01-01', '2022-01-03') #проверка когда всё ок
```

```python
date_range('2022-011', '2022-01-03') #неверный формат первой даты
```

```python
date_range('2022-01-01', '2022-013') #неверный формат второй даты
```

```python
date_range('2022-01-11', '2022-01-03') #дата 1 > дата 2
```
