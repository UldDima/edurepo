```python
import pandas as pd
```

```python
data = pd.read_excel('h3d.xlsx')
```

```python
data
```

```python
tab1 = data[['datekey', 'RezonItemID', 'GMV', 'Items']]
tab2 = data[['BrandID', 'RezonItemID.1', 'ItemGroupID']]
tab3 = data[['ID', 'Name', 'Type']]
```

```python
tab1.dropna()
```

```python
tab2 = tab2.dropna()
tab2
```

```python
tab3 = tab3.dropna()
tab3
```

```python
tab1['datekey'] = pd.to_datetime(tab1['datekey'], format='mixed')
```

```python
tab1 = tab1[(tab1['GMV'] > 100) & (tab1['Items'] > 0) & (tab1['GMV']/tab1['Items']<300000)]
```

```python
tab1
```

```python
ptype = tab3[(tab3['Type'] == 'ProductType')]
bname = tab3[(tab3['Type'] == 'BrandName')]
```

```python
ptype
```

```python
bname
```

```python
data2 = pd.merge(tab2, bname, left_on='BrandID', right_on='ID', how='left')
```

```python
data2 = pd.merge(data2, ptype, left_on='ItemGroupID', right_on='ID', how='left')
```

```python
data2
```

```python
val = tab1[['datekey', 'RezonItemID']].groupby(tab1[['datekey', 'RezonItemID']].columns.tolist(),as_index=False).size()
val['size'].unique()
```

```python
tab4 = tab1.groupby(['datekey', 'RezonItemID']).agg('sum').reset_index()
tab4
```

```python
data2 = pd.merge(data2, tab4, left_on='RezonItemID.1', right_on='RezonItemID', how='right')
```

```python
data2
```

```python
data3 = data2[['datekey', 'RezonItemID', 'GMV', 'Name_x', 'ItemGroupID', 'Name_y']].rename(columns={"Name_x": "Brand Name", "Name_y": "ProductType"})
data3
```

```python
data3.to_excel("output.xlsx")
```
